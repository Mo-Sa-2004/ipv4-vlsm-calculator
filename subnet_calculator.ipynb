# IPv4 VLSM Calculator — demo

This notebook shows `allocate_vlsm` from `subnet_calculator.py` being used directly as a library, with the result rendered as a `pandas` table.

For the interactive command-line version (and the full explanation of why requests are allocated largest-first), see `subnet_calculator.py` and the README.

```python
import ipaddress

import pandas as pd

from subnet_calculator import allocate_vlsm
```

```python
base = ipaddress.IPv4Network("192.168.1.0/24")

# Requested in a deliberately unsorted order -- allocate_vlsm sorts
# internally, so this still comes back with no overlaps.
requested_hosts = [20, 50, 2, 10]

subnets = allocate_vlsm(base, requested_hosts)

pd.DataFrame(
    [
        {
            "Requested hosts": sub.requested_hosts,
            "Network": str(sub.network),
            "Usable range": f"{sub.first_usable} - {sub.last_usable}",
            "Usable hosts": sub.usable_hosts,
        }
        for sub in subnets
    ]
)
```
